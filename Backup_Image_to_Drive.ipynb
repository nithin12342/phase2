# Backup Azure Docker Image to Google Drive (via Colab)

Since downloading a 5GB Docker image locally on a slow connection is error-prone, we can use **Google Colab's Gigabit internet** to transfer the image directly between Azure Container Registry (ACR) and Google Drive using `skopeo`.

### Instructions

1.  Run the cell below. It will mount your Drive and perform the transfer in minutes.
2.  To restore later on Docker Desktop, run: `docker load -i /path/to/h5-backend-v10.tar`.

```python
# 1. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# 2. Install Skopeo (Tool to copy container images without Docker Daemon)
!sudo apt-get update
!sudo apt-get install -y skopeo

# 3. Define Credentials & Registry
ACR_NAME = "h5regmanual1653298546"
ACR_LOGIN_SERVER = f"{ACR_NAME}.azurecr.io"

# --- FILL THESE IN ---
ACR_USERNAME = "h5regmanual1653298546"
ACR_PASSWORD = "7zxhuyuatqWo31VRLIwCVdKjKKrLmzl1freLa3VRWgXTICkHhK7TJQQJ99CBACqBBLyEqg7NAAACAZCRdi5k"

# 4. Login to Azure Registry
!skopeo login {ACR_LOGIN_SERVER} -u {ACR_USERNAME} -p {ACR_PASSWORD}

# 5. Transfer Image to Google Drive (.tar archive)
IMAGE_TAG = "v10"
SOURCE_IMAGE = f"docker://{ACR_LOGIN_SERVER}/h5-backend:{IMAGE_TAG}"
DESTINATION_TAR = f"docker-archive:/content/drive/MyDrive/h5-backend-{IMAGE_TAG}.tar:h5-backend:{IMAGE_TAG}"

print("Starting transfer from Azure to Google Drive. This may take a few minutes...")
!skopeo copy {SOURCE_IMAGE} {DESTINATION_TAR}

print("✅ Transfer Complete! The Docker image is safely stored in your Google Drive.")
```
